```python
import corner
import numpy as np
import ast
```

```python
model = "CGal_4_70_0.01_E_nau_single"
samples = np.load(f"/home/user/THESIS/getting_plots/{model}/nautilus_samples.npy")
weights = np.load(f"/home/user/THESIS/getting_plots/{model}/nautilus_weights.npy")
params = np.load(f"/home/user/THESIS/getting_plots/{model}/nautilus_params.npy")
print(params)
with open(f"/home/user/THESIS/getting_plots/{model}/nautilus_ndim.txt") as f:
    ndim = int(f.read())
with open(f"/home/user/THESIS/getting_plots/{model}/nautilus_labels.txt") as f:
    labels_str = f.read()
    cleaned = labels_str.strip().strip('[]')  # remove surrounding brackets
    labels = [label.strip().strip("'\"") for label in cleaned.split(',')]

print(labels)

```

```python
quantiles = [0.16,0.50,0.84]
cfig = corner.corner(samples, bins = 60, weights=weights, title_quantiles=quantiles,quantiles=quantiles,show_titles=True,
                     title_kwargs={"fontsize": 16}, label_kwargs={"fontsize": 16}, tick_params ={"labelsize":14}, labels=labels, color='purple',plot_datapoints=True, 
                     range=np.repeat(0.999,ndim))

#cfig.savefig(f'PLOTS/{model}_corner.pdf',bbox_inches='tight')
```
